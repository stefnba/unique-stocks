```python
import requests
```

```python
r = requests.post(
    "http://localhost:8080/v1/statement",
    data="SELECT * FROM iceberg.tiny_seed_data.countries",
    headers={"X-Trino-User": "admin"},
)

r.raise_for_status()

result = r.json()
result
```

```python
from trino.dbapi import connect

conn = connect(
    host="localhost",
    port=8080,
    user="admin",
    # catalog="<catalog>",
    # schema="<schema>",
)
cur = conn.cursor()
cur.execute("SELECT * FROM iceberg.tiny_seed_data.countries LIMIT 2")
rows = cur.fetchall()

desc = cur.description
column_names = [col[0] for col in desc]
column_names

data = [dict(zip(column_names, row)) for row in rows]

print(data)
```
